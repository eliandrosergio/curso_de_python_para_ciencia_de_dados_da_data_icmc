## Computação Vetorizada

### Aritmética

````python
arr_1 = np.arange(5)
arr_2 = np.arange(5, 10)

print(arr_1, arr_2, sep='\n')
````

````python
# Operação entre arrays
print(arr_1 + arr_2)
print(arr_1 - arr_2)
print(arr_1 / arr_2)
print(arr_1 * arr_2)
````

````python
# Operação com escalar
print(arr_1 + 2)
print(arr_1 - 2)
print(arr_1 * 2)
print(arr_1 / 2)
````

### Comparações

````python
arr_1 = np.array([1, 2, 3, 4, 5, 6])
arr_2 = np.array([1, 2, 5, 4, 3, 6])

print(arr_1, arr_2, sep='\n')
````

````python
# comparação entre arrays
print(arr_1 < arr_2)
print(arr_1 > arr_2)
print(arr_1 == arr_2)
print(arr_1 != arr_2)
````

### Manipulando os elementos do array com ufuncs e fatiamento

````python
# manipulando os elementos do array com ufuncs (fnuções universais)
a = np.array([10, 20, 30, 40])
print(a)

# adicionar ao final
a = np.append(a, 50)
print(a)

# adicionar em uma posição específica
a = np.insert(a, 1, 99)
print(a)

# remover em uma posição específica
a = np.delete(a, 1)
print(a)

# remover o último
a = np.delete(a, -1)
print(a)
````

````python
# manipulando os elementos do array com fatiamento (slicing)
a = np.array([10, 20, 30, 40])
print(a)

# adicionar ao final
a = np.array([*a, 50])
print(a)

# adicionar em uma posição específica
pos = 1 # posição 1
a = np.array([*a[:pos], 99, *a[pos:]])
print(a)

# remover em uma posição específica
pos = 1 # posição 1
a = np.array([*a[:pos], *a[pos+1:]])
print(a)

# remover o último
a = a[:-1]
print(a)
````

#### Exercício 3
Gere um array ```A``` contendo todos os números ímpares de 0 a 100. A partir de ```A```, crie outro array que contenha a soma cumulativa dos elementos de ```A``` que são divisíveis por 3 e 5 simultaneamente.

````python
# gerando o array 'A' com os números ímpares de 0 a 100
A = np.arange(1, 101, 2)

# criando o array com a soma cumulativa dos elementos de 'A'
# que são divisíveis por 3 e 5 simultaneamente 
B = A[(A % 3 == 0) & (A % 5 == 0)].cumsum()

# Exibindo o array da soma cumulativa
print(B)
````

#### Exercício 4
Faça uma função que dado dois vetores (valores e pesos), calcule a média ponderada de valores com seus respectivos pesos.

Observação: não é permitido utilizar a função np.average do NumPy.

````python
# criamos a função primeiro
# média ponderada é igual a soma da multiplicação entre valores e pesos, dividindo pela soma de pesos
def media_ponderada(valores, pesos):
    return (valores * pesos).sum() / pesos.sum()

# preparamos os dois vetores
valores = np.array([1, 2, 5])
pesos = np.array([3, 2, 1])

# exibimos o resultado
print(media_ponderada(valores, pesos))
````

Considere um array bidimensional no qual cada vetor de uma dimensão possui 3 pontuações que um jogador recebeu em 3 provas distintas. Faça uma função que dado esse array, retorne a prova com maior soma de pontuações.

````python
# criamos a função primeiro, devemos somar os valores na vertical (colunas)
# vamos usar a ufunc sum() com o parâmetro axis=0
def prova_com_maior_pontuacao(pontuacoes):
    return np.sum(pontuacoes, axis=0).max()

# preparamos o array com as pontuações
pontuacoes = np.array([[6.5, 7.8, 5.4], [9.5, 8.8, 9.1], [4.5, 7.3, 5.9]])

# exibimos a prova com maior soma de pontuações
print(prova_com_maior_pontuacao(pontuacoes))
````

#### Exercício 6
Você é um professor e está organizando as notas de seus alunos. Para tanto, você precisa realizar algumas tarefas, como:

- Determinar a média das notas
- Determinar a maior e a menor nota e os alunos que as tiraram
- Determinar as notas dos alunos aprovados e reprovados

Considere que são aprovados os alunos com notas maiores ou iguais a 5.0.

````python
# preparamos o nosso array com as notas
print('Todas as notas de todos os alunos')
notas = np.array([2.5, 7.8, 6.7, 9.4, 3.8, 5.1, 4.4, 8.2])

notas_list = notas.tolist()
for pos, nota in enumerate(notas_list):
    print(f'{pos}º aluno: {nota} pontos')

# realizamos a primeira tarefa: Determinar a média das notas
print('\n1º Tarefa')
media_das_notas = notas.sum() / len(notas)
print('As notas possuem uma média de: ', media_das_notas)

# realizamos a segunda tarefa: Determinar a maior e a menor nota e os alunos que as tiraram
# nota que adicionamos um '+ 1' aos índices da maior e menor nota nas suas exibições
# isso acontece apenas para fins de estética
print('\n2º Tarefa')
maior_nota = notas.max()
indice_maior_nota = notas.argmax()
print(f'O {indice_maior_nota + 1}º aluno tirou a maior nota, de {maior_nota} pontos')

menor_nota = notas.min()
indice_menor_nota = notas.argmin()
print(f'O {indice_menor_nota + 1}º aluno tirou a menor nota, de {menor_nota} pontos')

# realizamos a terceira tarefa: Determinar as notas dos alunos aprovados e reprovados
# Considere que são aprovados os alunos com notas maiores ou iguais a 5.0
print('\n3º Tarefa')
alunos_aprovados = notas[notas >= 5.0]
alunos_reprovados = notas[notas < 5.0]

print('Notas dos alunos aprovados:', alunos_aprovados)
print('Notas dos alunos reprovados:', alunos_reprovados)
````

#### Exercício 7
Um quadrado mágico é todo quadrado de lado ```n``` em que os inteiros 1 a ```n²``` aparecem sem repetição e em que as somas dos elementos das linhas, colunas e diagonais são iguais.

Faça uma função que receba uma matriz quadrada e que verifique se ela é um quadrado mágico ou não.

Exemplo: O quadrado ```[2 9 4] [7 5 3] [6 1 8]``` é um quadrado mágico. Observe que as linhas, colunas e diagonais possuem soma igual a 15.

````python
# função que diz se uma matriz quadrada é um quadrado mágico ou não
def ehQuadradoMagico(matriz):
    # verificamos se a matriz recebida é um quadrao
    # o valor da altura (y) precisa ser o mesmo da largura (x)
    if matriz.shape[0] != matriz.shape[1]:
        return False

    # calculamos as somas da diagonal e da anti-diagonal
    soma_das_diagonais = np.array([np.trace(matriz), np.trace(np.fliplr(matriz))])

    # calculamos as somas das linhas e das colunas
    soma_das_linhas = np.sum(matriz, axis=1)
    soma_das_colunas = np.sum(matriz, axis=0)

    # verificando se todos os elementos das somas são iguais
    # segundo a minha lógica matemática, se os valores mínimo e máximo forem iguais
    # então todos os valores do vetor são iguais
    verificacao = ((soma_das_linhas.max() == soma_das_linhas.min()) and 
                   (soma_das_colunas.max() == soma_das_colunas.min()) and
                   (soma_das_diagonais.max() == soma_das_diagonais.min()))

    # retormos o resultado da verificação
    # pois ele já diz se o matriz é um quadrado mágico ou não
    return bool(verificacao)
````

````python
# função auxiliar para pegar
# que pega o resultado booleano de ehQuadradoMagico() e exibe uma mensagem adequada
def printEhQuadradoMagico(matriz):
    if ehQuadradoMagico(matriz):
        print('É um quadrado mágico!')
    else:
        print('Não é um quadrado mágico!')

# criamos os quadrados
quadradoMagico = np.array([[2, 7, 6], [9, 5, 1], [4, 3, 8]])
quadradoMagico2 = np.array([[2, 9, 4], [7, 5, 3], [6, 1, 8]])
quadradoNaoMagico = np.array([[1, 2, 3], [5, 6, 4], [7, 9, 8]])

# verificamos se são quadrados mágico ou não
printEhQuadradoMagico(quadradoMagico)
printEhQuadradoMagico(quadradoMagico2)
printEhQuadradoMagico(quadradoNaoMagico)
````
