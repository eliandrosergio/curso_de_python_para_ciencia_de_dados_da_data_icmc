# Matplotlib

É uma lib do Python que permite fazer plotagem de gráficos por meio de dados matemáticos.

Ele possui dois objetos fundamentais, as **Figures** e os **Axes**.

- **Figures**: são o "corpo" da sua plotagem. Ele pode vir em uma janela externa, ou dentro de um notebook.
- **Axes**: dentro das *Figures* tem os *Axes*, que são o "conteúdo" de uma *Figure*.

````python
# primeiro de tudo devemos importar a lib com o alias 'plt'
import matplotlib.pyplot as plt

# importando também o NumPy com o alias np para auxiliar
import numpy as np
````

## Criando Figures e Axes

````python
# podemos criar uma Figure com o método figure()
# onde 'str' é a unidade de dimensão que pode ser 'px', 'cm' ou 'in' (o padrão)
fig = plt.figure(figsize=(4, 4), facecolor='pink', layout='tight')
fig.suptitle('Título da Figure')

# e podemos criar um Axes, para uma Figure, com o método .add_subplot(), usando nesta mesma Figure
# add_subplot(número de linhas, número de colunas, índice)
ax1 = fig.add_subplot(1, 2, 1)
ax1.set_title('Título do Axes 1')

ax2 = fig.add_subplot(2, 2, 2)
ax2.set_title('Título do Axes 2')

ax3 = fig.add_subplot(2, 2, 4)
ax3.set_title('Título do Axes 3')
````

````python
# também podemos criar a Figure e o Axes de uma vez com um único comando
# para isso usamos o método subplots() que retorna a Figure e os Axes
fig, axs = plt.subplots(2, 2, figsize=(4, 4), layout='tight', facecolor='cyan')
axs[0, 0].set_title('Título do Axes 1')
axs[0, 1].set_title('Título do Axes 2')
axs[1, 0].set_title('Título do Axes 3')
axs[1, 1].set_title('Título do Axes 4')
````

````python
# como normalmente um único Axes é o mais comum
# o jeito mais rapido para isso é usando plt.subplots() sem parâmetro algum
fig, ax = plt.subplots()

# ou talvez com apenas o parâmetro do tamanho
# o figsize recebe uma tupla, que pode ser do tipo (num, num) ou (num, num, str)
fig, ax = plt.subplots(figsize=(4, 4))
````

### plt.stem()
O plt.stem() é um meio-termo entre o plot() e o scatter(); ele desenha um ponto (um círculo) para cada valor x-y e, além disso, liga cada ponto ao eixo x com uma linha vertical, como se fosse um pirulito. 

````python
# número de pães vendidos numa cantina em cada dia da semana
dias = [1, 2, 3, 4, 5, 6, 7]
paes = [5, 8, 3, 9, 6, 12, 4]
plt.stem(dias, paes)
````

Agora vamos aprender alguns parâmetros legais:

```plt.stem(x, y, linefmt=None, markerfmt=None, basefmt=None, bottom=0)```

- **linefmt**: estilo ou cor da linha vertical, por exemplo 'grey' ou '--'.
- **markerfmt**: forma do ponto: 'o', 'D', 's'.
- **basefmt**: estilo da linha base, onde as hastes começam.
- **bottom**: a altura de onde as hastes partem. O padrão é 0.

````python
# temperatura em relação à média (25 °C)
dias = [1, 2, 3, 4, 5]
temp = [24, 27, 22, 29, 25]
plt.stem(dias, temp, linefmt='--', markerfmt='D', basefmt='grey', bottom=25)
````

### plt.stairs()

O plt.stairs() desenha um gráfico em degraus, como uma escada. Diferente do plot(), aqui mandamos apenas uma lista, a dos valores; ele cria sozinho as posições no eixo x (0, 1, 2, 3...) e desenha cada valor como um trecho horizontal, ligando um ao outro com linhas verticais.

````python
# preço do pão (em Kz) ao longo de 6 meses
preco_pao = [100, 100, 120, 120, 150, 150]
plt.stairs(preco_pao)
````

Agora vamos aprender alguns parâmetros legais:

```plt.stairs(values, edges=None, baseline=0, fill=False)```

- **values**: a altura de cada degrau.
- **edges**: onde cada degrau começa e acaba. Tem sempre um elemento a mais do que values. Se não mandares, ele usa 0, 1, 2...
- **fill=True**: pinta a área debaixo da escada.

````python
# preço da corrida por distância
preco = [500, 800, 1200]
limites_km = [0, 5, 10, 20]

# poderia ser assim: plt.stairs(preco, edges=limites_km, fill=True)
# mas vamos usar o método simples, sem a palavra 'edges' mas ainda com os valores (também funciona)
plt.stairs(preco, edges=limites_km, fill=True)
````

### plt.pie()

O plt.pie() cria o gráfico de pizza (ou de torta). Aqui mandamos só uma lista de valores x, e ele calcula sozinho a percentagem de cada fatia, ou seja, quanto cada valor representa do total.

````python
# onde vai o dinheiro do mês (em Kz)
gastos = [45000, 30000, 15000, 10000]
nomes = ['Renda', 'Comida', 'Transporte', 'Internet']
plt.pie(gastos, labels=nomes)
````

Agora vamos aprender alguns parâmetros legais:

```plt.pie(x, explode=None, labels=None, colors=None, autopct=None, startangle=0)```

- **labels**: lista de nomes, um por fatia.
- **autopct**: formato da percentagem escrita na fatia, por exemplo '%1.1f%%' (uma casa decimal).
- **startangle**: ângulo onde a primeira fatia começa. Com 90 começa no topo.
- **explode**: lista que afasta fatias do centro. [0, 0.1, 0] afasta a segunda.
- **colors**: lista de cores, uma por fatia.

````python
# como os colegas vêm para a escola
meios = ['A pé', 'Táxi', 'Autocarro', 'Carro']
pessoas = [4, 8, 6, 2]
plt.pie(pessoas, labels=meios, autopct='%1.1f%%', startangle=90,
        explode=[0, 0.1, 0, 0], colors=['Orange', 'Pink', 'Green', 'Cyan'])
````

São 20 pessoas, então as fatias são **20%**, **40%**, **30%** e **10%**. O ```explode``` destaca o táxi, que é a maior.

### plt.hist()

O plt.hist() cria o histograma. Ele parece o gráfico de barras, mas é diferente; no bar() cada barra é um identificador (um nome, um mês), já no hist() nós mandamos uma lista com muitos valores soltos e ele agrupa esses valores em intervalos (chamados de bins), contando quantos valores caem em cada um.

````python
# as notas de 20 alunos numa turma (de 0 a 20)
notas = [8, 9, 10, 11, 12, 12, 13, 13, 13, 14, 14, 14, 14, 15, 15, 16, 16, 17, 18, 19]
plt.hist(notas, bins=[8, 11, 14, 17, 20])
````

O parâmetro ```bins``` diz onde ficam as divisões; aqui temos 4 intervalos: **8 a 11**, **11 a 14**, **14 a 17** e **17 a 20**. Cada intervalo inclui o início mas não o fim; só o último inclui os dois. Se não escrevermos o bins, ele usa 10 intervalos por padrão.

Agora vamos aprender alguns parâmetros legais:

```plt.hist(x, bins=10, range=None, density=False, rwidth=None, edgecolor=None)```

- **bins**: um número (quantos intervalos iguais) ou uma lista com os limites exatos.
- **edgecolor**: cor da borda das barras. Ajuda a ver onde cada uma acaba.

````python
# alturas de 12 pessoas
alturas = [1.52, 1.60, 1.63, 1.65, 1.68, 1.70, 1.70, 1.72, 1.75, 1.78, 1.80, 1.85]
plt.hist(alturas, bins=5, edgecolor='black')
````

Com ```bins=5``` o Matplotlib divide o espaço entre o menor e o maior valor em 5 partes iguais. Experimenta ```bins=3``` e ```bins=10``` para ver como a forma muda.

### plt.ecdf()

````python
# o tempo de espera (em minutos) de 8 pessoas por um táxi na paragem
espera = [5, 8, 10, 12, 12, 15, 20, 30]
plt.ecdf(espera)
````

Como são 8 pessoas, cada uma vale ```1/8 = 0.125```; por isso em 5 minutos o gráfico chega a **0.125**, em 8 a **0.25**, em 10 a **0.375**, e em 12 salta para **0.625** (porque 2 pessoas esperaram 12). Assim, **62.5%** das pessoas esperaram até **12** minutos. Essa função existe só a partir do Matplotlib 3.8.

Agora vamos aprender alguns parâmetros legais:

```plt.ecdf(x, weights=None, complementary=False, orientation='vertical')```

- **complementary=True**: inverte a pergunta e mostra a fração que é maior do que cada valor.
- **orientation**: 'vertical' ou 'horizontal'.

````python
# quantos esperaram mais de 15 minutos
espera = [5, 8, 10, 12, 12, 15, 20, 30]
plt.ecdf(espera, complementary=True, orientation='horizontal')
````

Agora a linha desce em vez de subir, e vai da esquerda para a direita (horizontal) em vez de cima para baixo (vertical).

Em ```x = 15``` o valor é ```0.25```, ou seja, **25%** das pessoas esperaram mais de **15** minutos.

### plt.imshow()

O plt.imshow() é diferente de todos os outros; ele não usa listas de x e y, mas uma matriz (uma lista de listas, ou um array do NumPy) e transforma cada número em um quadradinho colorido. Quanto maior o número, mais “forte” é a cor. Isso é o que chamamos de mapa de calor, e também é assim que o computador mostra imagens, porque uma foto é só uma matriz de números.

````python
# as vendas de 3 lojas (as linhas) durante 4 dias (as colunas)
Z = np.array([[2, 4, 6, 8], [1, 3, 5, 7], [9, 7, 5, 3]])
plt.imshow(Z, cmap='Blues')
````

O quadradinho ```Z[i][j]``` fica na linha ```i``` e coluna ```j```, e a linha ```0``` fica no topo. O parâmetro ```cmap``` escolhe a paleta de cores do mapa, também podes usar ```'Greens'```, ```'Reds'```, ```'Oranges'```, ```'Greys'```.

````python
# escrever o número dentro de cada quadrado
Z = np.arange(12).reshape(3, 4)
plt.imshow(Z, cmap='Blues')
plt.colorbar()

for i in range(3):
    for j in range(4):
        plt.text(j, i, Z[i][j], ha='center', va='center')
````

O ```np.arange(12)``` cria os números de **0** a **11** e o ```reshape(3, 4)``` arruma-os em **3** linhas por **4** colunas.

No ```plt.text``` o ```x``` é a coluna ```(j)``` e o ```y``` é a linha ```(i)```, por isso aparecem trocados.

O ```plt.colorbar()``` desenha a barra de cores, que serve de legenda para sabermos que cor corresponde a que número

## Customização

### Legendas
Devemos usar, na hora de plotar, o parâmetro ```label``` para definir legendas para os dados plotados, e usamos o método ```.legend()``` para mostrar esse label (legenda) no Axes.

````python
fig, ax = plt.subplots()
X = np.linspace(0, 10, 100)
Y1 = np.sin(X)
Y2 = np.cos(X)

ax.plot(X, Y1, label='Seno de X')
ax.plot(X, Y2, label='Cosseno de X')
ax.legend()
````

### Ticks: Xticks e Yticks
Ticks são os númerozinhos com traços que aparecem nos eixos x e y. O Matplotlib por padrão usa ticks pares, se pretendes alterar os ticks de um Axes podes usar os seguintes métodos:

- **set_xticks()**: para os ticks do eixo x.
- **set_yticks()**: para os ticks do eixo y.

````python
fig, ax = plt.subplots()
X = [1, 5, 9, 13]
Y = [8, 3, 6, 12]

ax.set_xticks([1, 5, 9, 13])
ax.set_yticks([8, 3, 6, 12])
ax.stem(X, Y)
````

### Cores
É possível mudar a cor de cada linha ou ponto individualmente.

````python
fig, ax = plt.subplots()
X = [1, 2, 3, 4]
Y = [5, 5, 5, 5]

ax.plot(X, Y, c='green')
ax.set_xticks([0, 1, 2, 3, 4, 5])
ax.set_yticks([0, 5, 10])
````

````python
fig, ax = plt.subplots()
X = np.random.rand(500)-0.5
Y = np.random.rand(500)-0.5
cores = np.sqrt(X**2 + Y**2)

pontos = ax.scatter(X, Y, c=cores, cmap='rainbow')
fig.colorbar(pontos)
````
