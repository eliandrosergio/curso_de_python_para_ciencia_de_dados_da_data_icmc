### Series

Criando **Series** com Objetos do Python

### Dataframes (DF)

#### Criando DFs com estruturas do Python

````python
# podes criar um DataFrame com Tuplas, Arrays e Listas, até com dicionários
alunos = ("Elian", "Biva", "Gai", "Gota")
disciplinas = np.array(["Matemática", "Física", "Química", "Língua Portuguesa"])
notas_possiveis = [
    [7.0, 16.5, 18.5, 11.5],
    [4.0, 6.5, 8.5, 14.0],
    [12.5, 11.0, 6.5, 0.0],
    [17.0, 15.0, 18.5, 5.5],
]

# podendo indicar qual estrutura será dos índices e qual será das colunas
df = pd.DataFrame(notas_possiveis, index=alunos, columns=disciplinas)
print(df)
````

````python
# também podes criar um DataFrame com apenas os dados (valores)
notas = [[7.5, 16.5, 18.0, 9.0],
         [14.0, 6.5, 8.0, 16.0],
         [12.5, 11.5, 6.5, 2.5],
         [7.0, 15.0, 18.5, 5.5]]

# e ele gera os índices e colunas automaticamente, sendo do tipo numérico
df = pd.DataFrame(notas)
print(df)
````

Para ler arquivos com ```pd.read_excel()```. é preciso instalar a lib do arquivo a ser lido:

- **pyxlsb**, para arquivos ```.xlsb```
- **xlrd**, para arquivos ```.xls``` (arquivos antigos do Excel)

### Merge/Join

````python
# pegar o CSV dos carros e criar um novo DF para os países dos carros
carros = pd.read_csv('data/my/carros.csv', index_col='num_entrada')
paises = pd.DataFrame({
    'fabricante': ['eliantti', 'zonda'],
    'pais': ['Angola', 'Italia'],
    'ano': [2020, 1992]
})

# exibindo os DF que serão usados no merge
print(carros, paises, sep='\n\n', end='\n\n')
````

````python
# realizando uma operação de merge, onde o DF que chama o método é o da esquerda
resultado = carros.merge(
    paises, # DF da direita, também poderia ser 'right=paises'
    left_on='marca', # coluna-chave da esquerda, usada para o join (formar pares)
    right_on='fabricante', # coluna-chave da direita, usada para o join (formar pares)

    # o que acontece com que não tem par, 'outer' para incluir todos
    # poderia ser 'inner' que só deixa quem tem par
    how='outer',

    # sufixos da esquerda e direita, das colunas em ambos com o mesmo nome
    suffixes=('_carro', '_marca'),

    # adiciona a coluna '_merge' para auditória, que diz de onde veio cada linha
    # both (ambos), left_only (apenas da esquerda) e right_only (apenas da direita)
    indicator=True
)

# exibindo o resultado do merge
print(resultado)
````

### GroupyBy

````python
comentarios = pd.read_csv('data/my/comentarios.csv')
comentarios.index.name = 'id'
print(comentarios)
````

````python
# a) as_index=True (padrão): chave vira índice
print(comentarios.groupby("user_id", as_index=True).size(), end='\n\n')

# b) as_index=False: chave continua como coluna
print(comentarios.groupby("user_id", as_index=False).size())
````

````python
# a) sort=True (padrão): ordenado pela chave
print(comentarios.groupby("user_id", sort=True).size(), end='\n\n')

# b) sort=False: ordem de aparição (1, 3, 4, 2)
print(comentarios.groupby("user_id", sort=False).size())
````

````python
# dropna (o CSV não tem nulos, então injetamos um NaN numa cópia)
df_nan = comentarios.copy()
df_nan.loc[2, "user_id"] = np.nan
print(df_nan, end='\n\n')

# a) dropna=True (padrão): linha com NaN some
print(df_nan.groupby("user_id", dropna=True).size(), end='\n\n')

# b) dropna=False: NaN vira um grupo
print(df_nan.groupby("user_id", dropna=False).size())
````

````python
# criamos um grupo com o método .groupby
# vamos agrupar as linhas por 'user_id'
gb = comentarios.groupby('user_id')

# porém, por ele ser um método lazer (preguiçoso), ele faz o agrupamento e mas nada
# não retorna nenhum DF ou número, só faz isso quando pedimos a ele, usando métodos nele
print(gb)
````

````python
# como o número de grupos de criados
print(gb.ngroups, end='\n\n')

# os grupos criados
print(gb.groups, end='\n\n')

# e os elementos/linhas de um grupo específico
print(gb.get_group(2))
````

````python
# as primeira N linhas de cada grupo
print(gb.head(1), end='\n\n')

# as primeira N linhas de cada grupo
print(gb.tail(1), end='\n\n')
````

````python
# a posição de cada linha em seu respectivo grupo, começa com 0
print(gb.cumcount(), end='\n\n')

# resumo estatístico de cada grupo
print(gb.describe())
````

````python
print(gb.size(), gb.count(), sep='\n\n')
````

````python
print(gb.nunique(), gb['photo_id'].std(), sep='\n\n')
````

````python
print(gb['photo_id'].min(), gb['photo_id'].max(), sep='\n\n')
````

````python
print(gb['photo_id'].first(), gb['photo_id'].last(), sep='\n\n')
````

### Manipular elementos de um DataFrame

````python
# pegando e exibindo o DF paises
paises = pd.read_csv('data/my/paises.csv')
print(paises)
````

#### Linhas

##### Adicionando uma Linha no fim

````python
# adicionando uma nova linha em uma posição que não existe usando .loc[]
df = paises.copy()
df.loc[2] = ['zonda', 'França', 1992]
print(df)
````

````python
# adicionando uma nova linha em uma posição que não existe usando .loc[]
novos_paises = pd.DataFrame([
    ['zonda', 'França', 1998],
    ['missa', 'Rússia', 2004]],
    columns=['fabricante', 'pais', 'ano']
)

df = paises.copy()
df = pd.concat([paises, novos_paises]).reset_index(drop=True)
print(novos_paises, df, sep='\n\n')
````

##### Adicionando uma linha em uma posição específica

````python
# criamos um novo DF com os dados a serem adicionados
pos = 1
novo = pd.DataFrame([
    ['zonda', 'França', 1998]],
    columns=['fabricante', 'pais', 'ano']
)
print(novo, end='\n\n')

# cortamos o DF antigo na posição desejada e adicionamos o novo nesse intervalo
novo = pd.concat([paises.iloc[:pos], novo, paises.iloc[pos:]], ignore_index=True)
print(novo)
````

##### Remover linhas

````python
# remover um única linha por valor do índice
df = paises.copy()
print(df, end='\n\n')

df = df.drop(index=1)
print(df, end='\n\n')

# remover várias linhas por valores do índice
df = paises.copy()
print(df, end='\n\n')

df = df.drop(index=[0, 1])
print(df)
````

````python
# remover um única linha por posição do índice
df = paises.copy()
print(df, end='\n\n')

df = df.drop(index=paises.index[1])
print(df)
````

````python
# remover a última linha
df = paises.copy()
print(df, end='\n\n')

df = df.iloc[:-1]
print(df)
````

````python
# remover linhas repetidas (duplicatas)
df = pd.DataFrame([
    ['zonda', 'França', 1998]],
    columns=['fabricante', 'pais', 'ano'],
    index=pd.Index([1, 2])
)
print(df, end='\n\n')

df = df.drop_duplicates()
print(df)
````

````python
# remover linhas com valores nulos (NaN)
df = paises.copy()
df.at[1, 'pais'] = np.nan
print(df, end='\n\n')

df = df.dropna()
print(df)
````

````python
# por condição, mantém só quem passa no teste
df = paises.copy()
print(df, end='\n\n')

df = df[df['ano'] > 2000]
print(df)
````

### Exercícios

#### Exercício de Merge/Join

````python
# listas de frutas e seus países de origem
frutas = ["Maçã", "Laranja", "Ameixa", "Uva", "Mirtilo", "Melancia"]
paises_frutas = ["Brasil", "Espanha", "Portugal", "Itália", "EUA", "Brasil"]

df_frutas = pd.DataFrame({'Fruta': frutas, 'País': paises_frutas})
print(df_frutas, end='\n\n')

# lista de esportes e seus países de origem
esportes = ["Futebol", "Basquete", "Vôlei", "Tênis", "Natação"]
paises_esportes = ["Brasil", "EUA", "Brasil", "França", "EUA"]

df_esportes = pd.DataFrame({'Esporte': esportes, 'País': paises_esportes})
print(df_esportes)
````

````python
# inner join
inner_join_df = df_frutas.merge(df_esportes, how='inner', on='País')
print(inner_join_df)
````

````python
# outer join
outer_join_df = df_frutas.merge(df_esportes, how='outer', on='País')
print(outer_join_df)
````
